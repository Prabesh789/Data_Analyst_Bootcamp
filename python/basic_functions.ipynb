# Functions

```python
def first_func():
    print('First Function in Python !!')
```

```python
first_func()
```

```python
def number_squared(arg):
    print(arg**2)
```

```python
number_squared(4)
```

```python
def number_squared_cust(arg1, arg2):
    print(arg1**arg2)
```

```python
number_squared_cust(4, 3)
```

```python
# Arbitary argument (unknown arguments)

args_tuple = (5,3,1,2,8)
def number_args(*args):
    print(args[0]*args[1])
```

```python
number_args(*args_tuple)
```

```python
# keywords arguments
def number_squared_cust(arg1, arg2):
    print(arg1**arg2)
```

```python
number_squared_cust(arg1 = 6, arg2 = 2)
```

```python
def number_kwarg(**number):
    print('My number is: ' + number['integer'] + ' & the other number is ' + number['integer2'])
```

```python
number_kwarg(integer = '678', integer2 = '567')
```

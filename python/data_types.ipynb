```python
# Sets: Sets are similar to lists and tuples, but sets cannot have duplicate values. Values within the set cannot be accessed using the index because it is inordered.
```

```python
daily_pints = {1,2,3}
```

```python
type(daily_pints)
```

```python
daily_pints_log = {1,2,31,2,3,4,1,2,5,6,3,2}
```

```python
print(daily_pints_log)
```

```python
# The use cases for the sets can be in the comparison of the two values
```

```python
wifes_daily_pints_log = {1,3,5,7,3,24,5,7,3,2,0}
```

```python
# kind of union
print(daily_pints_log | wifes_daily_pints_log)
```

```python
# kind of intersection
print(daily_pints_log & wifes_daily_pints_log)
```

```python
# kind of A - B
print(daily_pints_log - wifes_daily_pints_log)
```

```python
# kind of compliement
print(daily_pints_log ^ wifes_daily_pints_log)
```

# For Loops 

```python
integers = [1,2,3,4,5]
```

```python
for number in integers:
    print(number)
```

```python
for number in integers:
    print('Yes !!!')
```

```python
integers = [1,2,3,4,5]
for i in integers:
    print(i+i)
```

```python
ice_cream_dict = {'name': 'Alex Freberg', 'weekly intake': 5, 'fav_ice_cream': ['MCC', 'Chocolate']}
```

```python
for cream in ice_cream_dict.values():
    print(cream)
```

```python
for key, value in ice_cream_dict.items():
    print(key,':',value)
```

# Nested For Loops

```python
flavors = ['Vanila', 'chocolate', 'Cookie Dough']
toppings = ['Hot Fudge', 'Oreos', 'Marshmallows']
```

```python
for one in flavors:
    for two in toppings:
        print(one, 'in top of', two)
```

# If Else Statements

```python
if 25 > 10:
    print('Yes')
```

```python
if 25 < 10:
    print('Yes')
else:
    print('No')
```

```python
if 25 < 10:
    print('Yes')
elif 25 < 30:
    print('elif yes')
else:
    print('No')
```

```python
if 25 < 10 or 1 < 2:
    print('Yes')
elif 25 < 30:
    print('elif yes')
else:
    print('No')
```

```python
print('Yes') if 10 > 30 else print('Nope')
```

```python
# Nested if statements
if 25 < 10 or 1 < 3:
    print('It worked!')
    if 10 > 3:
        print('This nested if statement worked!')
elif 25 < 20:
    print('elif wprked!')
elif 25 < 21:
    print('elif 2 worked!')
elif 25 < 40:
    print('elif 3 worked!')
elif 25 < 50:
    print('elif 4 worked!')
else:
    print('if did not work')
```

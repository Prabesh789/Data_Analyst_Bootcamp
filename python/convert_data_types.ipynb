# Converting Data Types

```python
num_int = 6
type(num_int)
```

```python
num_str = '8'
type(num_str)
```

```python
# String to int
num_str_convert = int(num_str)
type(num_str_convert)
```

```python
num_sum = num_int + num_str_convert
print(num_sum)
type(num_sum)
```

```python
list_type = [1,2,3]
type(list_type)
```

```python
# List to tuple
conv_tuple = tuple(list_type)
type(conv_tuple)
```

```python
# list to set
list_type = [1,2,3,3,2,1,3,4]
```

```python
conv_set = set(list_type)
type(conv_set)
```

```python
dict_type = {'name': 'Prabesh', 'age': 24}
type(dict_type)
```

```python
dict_type.items()
```

```python
dict_type.values()
```

```python
dict_type.keys()
```

```python
# dict key in list
list(dict_type.keys())
```

```python
# dict values in list
list(dict_type.values())
```

```python
# String to list
str = 'From Nepal'
list(str)
```

```python
# String to set
str = 'From Nepal'
set(str)
```
